# Kapitel 3 - Regression

#### Faktafrågor:

**1:** Regressionsproblem kan bero på att man inte har en kontinuerlig beroende variabel. Detta kan leta till att model inte kan skapa en regression med en "bra" fel marginal. 

**2:** 
- RMSE (*Root mean squared error*) - Är ett mått för att beräkna hur bra en model presterar. Den "straffar" modelen hårt vilket kan leda till bättre resultat.
- MSE (*Mean squared error*) - är en snabbare mindre "straff" process. Den är bra för att rangordna modeller eller ge oss en generall känsla över hur dom presterar.
- MAE (*Mean Absolute error*) - är den som man räknar ut snabbast med men beräknar inte lika nog grant.

Man kan säga att man offrar snabbhet för nogranhet beroende på vilken model man vill använda.

**3:** Om vi bara ska ranordna modeller och vi har RMSE och MSE att välja på så är det bättre att använda MSE. Den är snabbare för att den inte behöver använda roten ur vilket tar längre tid att beräkna. Därför är MSE bättre att använda i detta syfte.

**4:** Gradient descent är en optimeringsalgoritm som räknar ut lutningen för att minska felet i en modell. Man kan tänka sig en studsboll som släpps i ett kuperat landskap (där höjden är felet) och letar efter den lägsta punkten.

Problemet ligger i hur stora hopp vi låter den göra (learning rate): Om vi låter den göra för små hopp tar det en evighet eller så kan den fastna i ett lokalt minimum i stället för det globala minimumet. Om vi i stället gör för stora hopp kommer den att studsa okontrollerat och missa.

**5:** Bias variance trade-off är ett begräpp inom ML som handlar om hur mycket eller lite vi "styr" en modell.
Om den har hög bias blir den under anpassad och om den har hög variance blir den överanpassad. 

Med andra ord om vi gör våran modell för enkel kan den missa en massa värden den och blir för general. vilket inte ger oss bra resultat.
Och gör vi den för komplex kommer den bli för dålig på att anpassa sig efter ny data. 

lagom är bäst.

**6:** Olika regressions modeller:

Linjär: Drar en raklinje mellan x och y för att få ut ett medelvärde baserat på dom observationerna den har. 

Ridge: Är en modell där vi använder oss av Alpha för att kontrollera hur hårt vi "straffar våran model" genom att styra värdena på "vikterna". detta för att motarbeta variance och över anpassning

Lasso: Är lik Ridge men istället för att mindra så sätter den låga värden till 0.

Elastic net: är en kombination av Ridge och Lasso

SVM (Support vector machine): Om vi tänker oss att våran vanliga linjära regression är en stig så gör SVM om detta till en väg och kolla bara på alla värden som finns på vägen. vi kan ståra hur stor vägen och på så sätt inte kolla på värden som är outliers.

Decision three: är en regression som vi kontrollerar med hjälp av hyperparametrar för att få fram en så optimal model som möjligt. den tränar data som om det vore ett träd. först har vi rotnoden som går till inre noder för att tillslut komma till löv noderna.

Ensemble Learning: Är att den försöker kombinera olika modeller för att försöka få fram en modell som presterar bättre än vad den gjorde innan.

Random forest: är en ensamble av massa decision threes.

**7:** Whiteboxing är modeller som man lätt kan förklara varför den gör som den gör och blackboxing kan vi inte riktigt det för. eller mer exakt vad dom händer under huven så att säga.

**8:** Bagging och pasting användes för att slumpa ut data från ett dataset som man sen tar och gör ett nytt dataset som man kan träna på.
- Bagging: är att den slumpmässiga datan kan återanvändas
- Pasting: är att den slumpmässiga datan bara används en gång.

Bagging förekommer när man gör en random forest.
