**1:** Checklist. Dom sju stegn

- *Definera problem och skapa en helhets bild.* Vad är det vi vill att våran modell ska hjälpa oss att lös? vad är planen att den ska användas till?
- *Få tillgång/skaffa data.* Vi försöker få tag i relevant data som vi kan använda för att träna våran modell.
- *Utforska datan (EDA).* I detta går vi igen datan som vi har. Ser vi några mönster. Saknas det värden? Är datan komplett?
- *Bearbeta datan.* Nu är det dags att bearbeta datan genom att ev fylla i saknad data eller ta bort den (data cleaning). Även transformera datan för att få ut våra hyperparametrar. Behöver vi göra någon feature engineering för att få variabler vi kan arbeta med? 
- *ML-modellering.* Det är här vi bestämmer vilka modeller vi ska använda och vi tränar dom för att se om vi får fram önskaded resultat.
- *Presentera Modellen ev fynd.* Om vi du har lyckats få fram en modell som vi tycker presterar bra, så är det nu vi presenterar det för kollegor, arbetsgivare, berörda parter
- *produktionssätting och övervakning.* Nu ska vi ta och produktionssätta modellen till det ända målet som den är till för. Vi måste med gämna mellan rum kontrollera modellen så att den fortsätter att producera som den ska. kan hända att data den ska jobba med ändras av någon anledning och då måste vi kanske ta fram en ny modell och träna om tidigare modeller på ny relevant data.
